```python
#Part 1
import pandas as pd
import numpy as np

DATA_PATH = "../../data/FRED"


```

```python
# PART 2. 
time = np.arange(1950,2020,10)

times = []
for v in time: 
    file = f"{DATA_PATH}/FRED_monthly_{v}.csv"
    times.append(pd.read_csv(file, parse_dates=["DATE"]))


joined = pd.concat(times)
joined.set_index("DATE")
```

```python
#Part 3

file2 = f"{DATA_PATH}/GDP.csv"
gdp = pd.read_csv(file2, parse_dates=["DATE"], index_col=["DATE"])
gdp

merged_gdp = pd.merge(joined, gdp, on="DATE", how="inner")
merged_gdp

#The observations in the merged dataset, is that going from dividin gthe rows by three. The key used is "DATE" which is now only reporter quartarly.
```

```python
df = pd.DataFrame()
df["CPI_avg"] = ((merged_gdp["CPI"].pct_change())*100).round(2)
df["GDP_avg"]= ((merged_gdp["GDP"].pct_change())*100).round(2)
df["FEDFUNDS_avg"]= merged_gdp["FEDFUNDS"].diff()
df["REALRATE_avg"]= merged_gdp["REALRATE"].diff()
df["UNRATE_avg"]= merged_gdp["UNRATE"].diff()
df["LFPART_avg"]= merged_gdp["LFPART"].diff()

df["CPI_avg"].mean().round(2)
```

```python
#Part 4.2
corr = df.corr()
corr["GDP_avg"]
corr

```

# Exercise 3: Weekly returns of the Magnificent Seven

In this exercise, you are asked to analyze the weekly stock market returns of the
so-called Magnificent Seven, which are some of the most successful tech companies
of the last few decades:
Apple (AAPL), Amazon (AMZN), Alphabet/Google (GOOGL), Meta (META), Microsoft (MSFT), Nvidia (NVDA), and Tesla (TSLA).

The data for this exercise is located in the folder `data/stockmarket/`.
Each stock file contains the following variables:

| Variable | Description |
|----------|-------------|
| `Date` | Trading date. |
| `Open` | Opening stock price for the trading day. |
| `Close` | Closing stock price for the trading day. |

```python
#PART 1. 

import pandas as pd
ticker = ["AAPL","AMZN","GOOGL","META","MSFT","NVDA","TSLA"]
DATA_PATH = "../../data/stockmarket/"

returns = []
files = []

for n in ticker: 
    file = pd.read_csv(f"{DATA_PATH}{n}.csv", parse_dates=["Date"],index_col=["Date"])
    files.append(file)
    weekly_re = file.resample("W")
    w_returns = ((weekly_re["Close"].last()-weekly_re["Open"].first())/weekly_re["Open"].first())*100
    returns.append(w_returns)
```

```python
#Part 2. 
df = pd.concat(returns, axis = 1, join ="inner", keys = ticker).iloc[1:-1] #Concat is automaticaly outer join
df.head(10)
```

```python
#Part 3. 

#Finding the correlation between the stocks
corr = df.corr()
corr.iloc[1,1]


```

```python
#Plotting a matrix
import pandas as pd
pd.plotting.scatter_matrix(df, alpha = 0.5, figsize=(10,5), grid=True, marker="." )
```

```python
import matplotlib.pyplot as plt

#The setup the lengt of the colns
cols = df.columns
n = len(cols) #7


#Creating the grid
fig, axes = plt.subplots(n,n, figsize=(10,10), sharex="col", sharey=True)


#Nesten loops for the plotting

for i, row_col in enumerate(cols): #enumerate gives both the number and the name (0,"AAPL"), (1, "AMZN")
    for j, col_col in enumerate(cols): 
        ax = axes[i,j]
        if i == j: 
            ax.hist(df[col_col].dropna(),bins=30)
        else: 
            ax.scatter(df[col_col],df[row_col],s=5,alpha=0.5, marker=".")
            ax.set_title(corr.iloc[i,j].round(2))
        if i == n - 1: ax.set_xlabel(col_col)   # bottom row
        if j == 0:     ax.set_ylabel(row_col)   # leftmost column
    
           




plt.tight_layout()   # adjusts spacing so nothing overlaps
plt.show()
```
